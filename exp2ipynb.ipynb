```python
def calculate_salary(basic, hra, da, allowance):
    """Calculates gross and net salary based on provided components."""

    # Gross Salary
    gross_salary = basic + hra + da + allowance

    # Deductions
    pf = basic * 0.12      # 12% PF
    tax = gross_salary * 0.10   # 10% Tax

    # Net Salary
    net_salary = gross_salary - (pf + tax)

    return {
        "gross_salary": gross_salary,
        "pf_deduction": pf,
        "tax_deduction": tax,
        "net_salary": net_salary
    }

# The original input prompt part is commented out for programmatic use
# basic = float(input("Enter Basic Salary: "))
# hra = float(input("Enter HRA: "))
# da = float(input("Enter DA: "))
# allowance = float(input("Enter Other Allowances: "))

# # Gross Salary
# gross_salary = basic + hra + da + allowance

# # Deductions
# pf = basic * 0.12      # 12% PF
# tax = gross_salary * 0.10   # 10% Tax

# # Net Salary
# net_salary = gross_salary - (pf + tax)

# # Output
# print("\n----- Salary Details -----")
# print("Gross Salary :", gross_salary)
# print("PF Deduction :", pf)
# print("Tax Deduction:", tax)
# print("Net Salary   :", net_salary)
```

```python
# Example Usage:
# Let's assume some values for basic, HRA, DA, and other allowances.
example_basic = 50000
example_hra = 10000
example_da = 5000
example_allowance = 2000

# Call the function with example values
salary_details = calculate_salary(
    example_basic, example_hra, example_da, example_allowance
)

# Print the results
print("\n----- Calculated Salary Details -----")
print(f"Basic Salary    : {example_basic}")
print(f"HRA             : {example_hra}")
print(f"DA              : {example_da}")
print(f"Other Allowances: {example_allowance}")
print(f"Gross Salary    : {salary_details['gross_salary']}")
print(f"PF Deduction    : {salary_details['pf_deduction']}")
print(f"Tax Deduction   : {salary_details['tax_deduction']}")
print(f"Net Salary      : {salary_details['net_salary']}")
```
