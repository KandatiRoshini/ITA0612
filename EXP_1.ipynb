```python
import java.util.Scanner;

class StudentGradeCalculator {
    public static void main(String[] args) {
        Scanner sc = new Scanner(System.in);

        int total = 0;
        double average;
        char grade;

        System.out.println("Enter marks for 5 subjects:");

        for (int i = 1; i <= 5; i++) {
            total += sc.nextInt();
        }

        average = total / 5.0;

        if (average >= 90)
            grade = 'A';
        else if (average >= 80)
            grade = 'B';
        else if (average >= 70)
            grade = 'C';
        else if (average >= 60)
            grade = 'D';
        else
            grade = 'F';

        System.out.println("Total Marks = " + total);
        System.out.println("Average Marks = " + average);
        System.out.println("Grade = " + grade);

        sc.close();
    }
}
```

```python
total = 0

print("Enter marks for 5 subjects:")

for i in range(1, 6):
    while True:
        try:
            mark = int(input(f"Enter mark for subject {i}: "))
            if 0 <= mark <= 100:
                total += mark
                break
            else:
                print("Please enter a mark between 0 and 100.")
        except ValueError:
            print("Invalid input. Please enter an integer for the mark.")

average = total / 5.0

if average >= 90:
    grade = 'A'
elif average >= 80:
    grade = 'B'
elif average >= 70:
    grade = 'C'
elif average >= 60:
    grade = 'D'
else:
    grade = 'F'

print("\nTotal Marks =", total)
print("Average Marks =", average)
print("Grade =", grade)
```
