```python
basic = 50000.0  # Example: Enter Basic Salary
hra = 20000.0    # Example: Enter HRA
da = 15000.0     # Example: Enter DA

gross = basic + hra + da

pf = basic * 0.12      # 12% PF
tax = gross * 0.10     # 10% Tax

net = gross - (pf + tax)

print("Gross Salary =", gross)
print("PF Deduction =", pf)
print("Tax Deduction =", tax)
print("Net Salary =", net)
```
